```python
# While LOOPS
""" While loop works on the basis of Condition.
Syntax:
starting point
While Condition:"""

```

```python
a = 0
while a < 10:
  print(a)
  a = a + 1
```

```python
a = 1
while a < 21:
  print(a)
  a = a + 1
```

```python
a = 10
while a >= 0:
  print(a)
  a = a - 1
```

```python
# Break
# Break the loop whenever break keyword is encountered.
# so it break the code whenever it meet the requirements.
for i in range(1,11):
  print(i)
  if i == 6:
    break
```

```python
a = 0
while a < 10:
  print(a)
  a = a + 1
  if a == 7:
    break
```

```python
#   Continue:
# Means: Stop the current iteration and move to the next one.
# In this example, we skip printing the number 6.
a = 0
while a < 10:
  a = a + 1
  if a == 6:
    continue
  print(a)
```

```python
a = 0
while a < 10:
  a = a + 1
  if a == 6:
    break
  print(a)
else:
  print("Break Not executed")
```
