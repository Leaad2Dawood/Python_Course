```python
# Question 4
# Overall, the code adds every whole number from 1 through the number you enter. For example, entering 5 calculates 1 + 2 + 3 + 4 + 5.
n = int(input("please tell which numbers you want the sum"))
s = 0
for i in range(1,n+1):
  s = s+ i
  print(f"your sum is {s}")
```

```python
# Even and odd Numbers
n= int(input("Please tell which number you want to Even or odd:- "))
even_sum = 0
odd_sum = 0
for i in range(1,n+1):
    if i%2 == 0:
        even_sum = even_sum + i
    else:
        odd_sum = odd_sum+ i
print(f"Hello your even sum is {even_sum} and your odd sum is {odd_sum}")
```

```python
a = int(input("please tell your Value :-"))
b = int(input("please tell your exponent :-"))
power = a
for i in range(b-1):
    power = power * a
print(f"After power your answer is {power}")
```

```python
n = int(input("give your number(Prime Check):-"))
count = 0
for i in range(1,n+1):
    if n % i == 0:
        count = count + 1
if count == 1:
        print("this is unique number")
elif count == 2:
        print("this is prime number")
else:
        print("Your number is Composite")
```

```python
n = int(input("give your number(Prime Check):-"))
for i in range(2,n):
    if n % i == 0:
        print("Sorry your number is composite")
        break
else:
        print("Your number is prime")
```
